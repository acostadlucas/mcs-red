```python
def count_columns_per_df(dfs, names=None):
    """
    Returns the number of columns (countries) in each DataFrame.
    """
    if names is None:
        names = [f'DataFrame {i+1}' for i in range(len(dfs))]
    
    summary = []
    for name, df in zip(names, dfs):
        summary.append((name, df.shape[1]))
    
    print("\nColumn Count Summary")
    print(f"{'DataFrame':<25} {'# Columns':>10}")
    print("-" * 40)
    for name, count in summary:
        print(f"{name:<25} {count:>10}")
```

```python
def compare_columns(df1, df2):
    # Get the set of column names from each DataFrame
    columns_df1 = set(df1.columns)
    columns_df2 = set(df2.columns)

    # Find columns that exist in both DataFrames
    in_both = columns_df1 & columns_df2

    # Find columns that exist only in df1
    only_in_df1 = columns_df1 - columns_df2

    # Find columns that exist only in df2
    only_in_df2 = columns_df2 - columns_df1

    # Return a dictionary with sorted lists of column names
    return {
        "in_both": sorted(in_both),
        "only_in_df1": sorted(only_in_df1),
        "only_in_df2": sorted(only_in_df2)
    }
```

```python
def get_year_ranges(list_of_dfs, names=None):
    """
    list_of_dfs: list of DataFrames with a 'Year' column
    names: optional list of names for each DataFrame
    """
    # If no names are provided, generate default names
    if names is None:
        names = [f"DataFrame_{i+1}" for i in range(len(list_of_dfs))]

    summary = []

    # Iterate through each DataFrame and its corresponding name
    for name, df in zip(names, list_of_dfs):
        if "Year" in df.columns:
            years = df["Year"].dropna()
            if not years.empty:
                summary.append({
                    "Indicator": name,
                    "Start Year": int(years.min()),
                    "End Year": int(years.max())
                })
            else:
                summary.append({
                    "Indicator": name,
                    "Start Year": None,
                    "End Year": None
                })
        else:
            summary.append({
                "Indicator": name,
                "Start Year": "❌ Missing 'Year' column",
                "End Year": "❌ Missing 'Year' column"
            })

    return pd.DataFrame(summary)
```

```python
def check_columns_in_filenames(dataframes, df_names, path):
    """
    Checks whether column names from each DataFrame exist in the filenames of a given path,
    accounting for differences like '-' in column names vs '_' in filenames.

    Parameters:
    - dataframes: list of pandas DataFrames
    - df_names: list of names corresponding to each DataFrame
    - path: directory path containing files

    Prints:
    - For each DataFrame, its name followed by the list of column names
      that do NOT appear in any of the filenames in the specified path.
    """
    try:
        filenames = os.listdir(path)
    except FileNotFoundError:
        print(f"❌ Path not found: {path}")
        return

    # Normalize filenames: lowercase and replace underscores with hyphens
    normalized_filenames = [f.lower().replace("_", "-") for f in filenames]

    for df, name in zip(dataframes, df_names):
        missing_columns = []
        for col in df.columns:
            normalized_col = col.lower().replace("_", "-")
            if not any(normalized_col in fname for fname in normalized_filenames):
                missing_columns.append(col)

        print(f"📊 DataFrame: {name}")
        if missing_columns:
            print(f"   ❌ Columns not found in filenames: {missing_columns}")
        else:
            print("   ✅ All columns found in filenames")
```

```python
def build_country_dataframes(macro_dict):
    sample_df = next(iter(macro_dict.values()))
    countries = [col for col in sample_df.columns if col != 'Year']
    country_dataframes = {}

    for country in countries:
        country_df = None

        for indicator, df in macro_dict.items():
            if country in df.columns:
                temp_df = df[['Year', country]].copy()
                temp_df.rename(columns={country: indicator}, inplace=True)
            else:
                # Create a NaN-filled column for missing country
                temp_df = pd.DataFrame({
                    'Year': df['Year'],
                    indicator: [float('nan')] * len(df)
                })

            country_df = temp_df if country_df is None else pd.merge(country_df, temp_df, on='Year', how='outer')

        country_df.sort_values(by='Year', inplace=True)
        country_dataframes[country] = country_df

    return country_dataframes
```

```python
def export_country_dataframes_to_parquet(country_dfs, output_path, label="macro_profile"):
    """
    Exports each country's DataFrame to a Parquet file with descriptive filenames.

    Parameters:
    - country_dfs: dict of country name -> DataFrame
    - output_path: string path to the folder where files will be saved
    - label: optional string to include in the filename for clarity
    """
    os.makedirs(output_path, exist_ok=True)

    for country, df in country_dfs.items():
        # Clean country name for file system
        safe_name = country.replace(" ", "_").replace("-", "_")
        file_name = f"{safe_name}_{label}.parquet"
        file_path = os.path.join(output_path, file_name)

        df.to_parquet(file_path, index=False)

    print(f"✅ Exported {len(country_dfs)} files to: {output_path}")
```
