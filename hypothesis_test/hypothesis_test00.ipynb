```python
def count_columns_per_df(dfs, names=None):
    """
    Returns the number of columns (countries) in each DataFrame.
    """
    if names is None:
        names = [f'DataFrame {i+1}' for i in range(len(dfs))]
    
    summary = []
    for name, df in zip(names, dfs):
        summary.append((name, df.shape[1]))
    
    print("\nColumn Count Summary")
    print(f"{'DataFrame':<25} {'# Columns':>10}")
    print("-" * 40)
    for name, count in summary:
        print(f"{name:<25} {count:>10}")
```

```python
def comparar_columnas(df1, df2):
    columnas_df1 = set(df1.columns)
    columnas_df2 = set(df2.columns)

    en_ambos = columnas_df1 & columnas_df2
    solo_en_df1 = columnas_df1 - columnas_df2
    solo_en_df2 = columnas_df2 - columnas_df1

    return {
        "en_ambos": sorted(en_ambos),
        "solo_en_df1": sorted(solo_en_df1),
        "solo_en_df2": sorted(solo_en_df2)
    }
```

```python
def obtener_rangos_de_anios(lista_dfs, nombres=None):
    """
    lista_dfs: lista de DataFrames con columna 'Year'
    nombres: lista opcional con nombres para cada DataFrame
    """
    if nombres is None:
        nombres = [f"DataFrame_{i+1}" for i in range(len(lista_dfs))]

    resumen = []

    for nombre, df in zip(nombres, lista_dfs):
        if "Year" in df.columns:
            anios = df["Year"].dropna()
            if not anios.empty:
                resumen.append({
                    "Indicador": nombre,
                    "Año Inicio": int(anios.min()),
                    "Año Fin": int(anios.max())
                })
            else:
                resumen.append({
                    "Indicador": nombre,
                    "Año Inicio": None,
                    "Año Fin": None
                })
        else:
            resumen.append({
                "Indicador": nombre,
                "Año Inicio": "❌ No tiene columna 'Year'",
                "Año Fin": "❌ No tiene columna 'Year'"
            })

    return pd.DataFrame(resumen)
```

```python
def eliminar_columnas(df, columnas_a_eliminar):
    """
    Elimina columnas específicas de un DataFrame.

    Parámetros:
    - df: DataFrame original
    - columnas_a_eliminar: lista de nombres de columnas a eliminar

    Retorna:
    - DataFrame sin las columnas especificadas
    """
    return df.drop(columns=columnas_a_eliminar, errors='ignore')
```
