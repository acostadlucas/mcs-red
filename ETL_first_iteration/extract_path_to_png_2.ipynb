```python
import os
import csv
```

```python
# Ruta base del proyecto
base_dir = os.getcwd()

# Ruta al directorio de plots
plots_dir = os.path.join(base_dir, 'plots_2')

# Ruta al directorio donde se guardará el CSV
data_dir = os.path.join(base_dir, 'data')
os.makedirs(data_dir, exist_ok=True)  # crea la carpeta si no existe
```

```python
# additional config
# Archivo CSV de salida
csv_path = os.path.join(data_dir, 'image_paths_2.csv')

# Lista para guardar los paths
image_paths = []
```

```python
# Recorrer todos los subdirectorios dentro de plots_2
for country_folder in os.listdir(plots_dir):
    country_path = os.path.join(plots_dir, country_folder)
    if os.path.isdir(country_path):
        for file in os.listdir(country_path):
            if file.endswith('.png'):
                full_path = os.path.join(country_path, file)
                image_paths.append(full_path)
```

```python
# Diccionario para guardar el conteo por país
conteo_por_pais = {}

# Recorrer cada subcarpeta en plots
for country_folder in os.listdir(plots_dir):
    country_path = os.path.join(plots_dir, country_folder)
    if os.path.isdir(country_path):
        # Contar archivos .png en la carpeta del país
        num_images = sum(
            1 for file in os.listdir(country_path)
            if file.lower().endswith('.png')
        )
        conteo_por_pais[country_folder] = num_images

# Imprimir resultados
print("📊 Conteo de imágenes por país:\n")
for pais, cantidad in conteo_por_pais.items():
    print(f"🌍 {pais}: {cantidad} imágenes")
```

```python
# Guardar los paths en un archivo CSV
with open(csv_path, mode='w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['image_path'])  # encabezado
    for path in image_paths:
        writer.writerow([path])

print(f"✅ Se guardaron {len(image_paths)} rutas en {csv_path}")
```

```python
def verificar_imagenes_faltantes(directorio_madre, indicadores):
    faltantes = []

    # Recorremos cada carpeta (país) dentro del directorio madre
    for nombre_pais in os.listdir(directorio_madre):
        ruta_pais = os.path.join(directorio_madre, nombre_pais)
        if not os.path.isdir(ruta_pais):
            continue  # Ignorar si no es carpeta

        archivos_png = [f for f in os.listdir(ruta_pais) if f.endswith('.png')]

        # Verificamos cada indicador
        for indicador in indicadores:
            esperado = f"{indicador}{nombre_pais}.png"
            if not any(esperado in archivo for archivo in archivos_png):
                faltantes.append((nombre_pais, esperado))

    return faltantes
```

```python
indicadores = [
    "Education_spending_",
    "Inflation_",
    "Share_of_industry_",
    "Tax_revenue_",
    "wb_government_effectiveness_",
    "wb_ruleoflaw_"
]
faltantes = verificar_imagenes_faltantes("plots_2", indicadores)

for pais, archivo in faltantes:
    print(f"Falta en {pais}: {archivo}")
```
